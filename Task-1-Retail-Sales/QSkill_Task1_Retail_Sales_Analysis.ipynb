### Objective
The aim of this project is to clean a retail sales dataset, understand the sales data and find useful business insights using Python, Pandas, Matplotlib and Seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

## 1. Load the Dataset

```python
url = "https://raw.githubusercontent.com/leonism/sample-superstore/master/data/superstore.csv"
df = pd.read_csv(url)

print("Rows and columns:", df.shape)
df.head()
```

## 2. Understand the Dataset

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate records:", df.duplicated().sum())
```

## 3. Data Cleaning

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Convert date columns into date format
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

# Remove rows where important values are missing
df = df.dropna(subset=['Sales', 'Profit', 'Quantity', 'Customer ID'])

print("Duplicate records after cleaning:", df.duplicated().sum())
print("Missing values after cleaning:")
print(df.isnull().sum().sum())
```

## 4. Basic Sales Statistics

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_quantity = df['Quantity'].sum()
total_customers = df['Customer ID'].nunique()
total_orders = df['Order ID'].nunique()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
print("Number of Customers:", total_customers)
print("Number of Orders:", total_orders)
```

## 5. Sales by Category

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
print(category_sales)

plt.figure(figsize=(7,5))
category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

## 6. Monthly Sales Trend

```python
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()
monthly_sales.index = monthly_sales.index.astype(str)

plt.figure(figsize=(12,5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## 7. Sales and Profit Relationship

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Sales', y='Profit', hue='Category', alpha=0.6)
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

## 8. Correlation Heatmap

```python
numeric_data = df[['Sales', 'Quantity', 'Discount', 'Profit']]

plt.figure(figsize=(7,5))
sns.heatmap(numeric_data.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## 9. Product Performance

```python
product_sales = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False)

print("Top 10 products by sales:")
print(product_sales.head(10))
```

## 10. Regional Performance

```python
region_data = df.groupby('Region').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity': 'sum'
}).sort_values('Sales', ascending=False)

region_data
```

## 11. Customer Performance

```python
customer_sales = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False)

print("Top 10 customers by sales:")
print(customer_sales.head(10))
```

## 12. Profit by Category

```python
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
print(category_profit)

plt.figure(figsize=(7,5))
category_profit.plot(kind='bar')
plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.xticks(rotation=0)
plt.show()
```

## 13. Observations and Business Insights

- The category with the highest sales can be given more attention because it contributes strongly to overall revenue.
- Sales change from month to month, so the business can use the monthly trend to plan inventory and promotions.
- The scatter plot helps to identify orders where high sales do not necessarily result in high profit.
- The heatmap shows the relationship between sales, quantity, discount and profit.
- Regional sales and profit are different, so a region with high sales should also be checked for profitability.
- Top customers can be targeted with offers and loyalty programs to encourage repeat purchases.
- Top-selling products can be kept well stocked, while low-profit products should be reviewed for pricing or discount decisions.

The analysis covered sales, profit, quantity, customers, products, categories and regions. The findings can help a retail business understand its better-performing areas and make better decisions about products, customers, regions and discounts.
